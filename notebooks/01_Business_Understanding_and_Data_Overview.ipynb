### **Library Imports**

Import the required Python libraries for data analysis.

```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

### **Olist Dataset Extraction**

Extract the uploaded ZIP file into a separate folder containing the nine CSV files for further data exploration.

```python
import zipfile
zip_path = "/content/archive.zip"
extract_path = "/content/olist_data"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("ZIP file extracted successfully!")
```

### **Dataset Overview**

Overview of the nine Olist CSV files.

```python
data_path = "/content/olist_data"

files = os.listdir(data_path)

for file in files:
    print(file)

print("\nTotal files:", len(files))
```

### **Dataset Dimensions**

Number of rows and columns in each dataset.

```python
for file in files:
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(data_path, file))
        print(f"{file}: {df.shape}")
```

### **Dataset Columns**

Column names available in each dataset.

```python
for file in files:
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(data_path, file), nrows=0)
        print(f"\n{file}")
        print(df.columns.tolist())
```

### **Dataset Sample Records**

Sample records from each dataset for initial data understanding.

```python
for file in files:
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(data_path, file), nrows=2)
        print(f"\n{'='*60}")
        print(file)
        display(df)
```

### **Missing Value Summary**

Missing-value counts for columns containing incomplete records.

```python
for file in files:
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(data_path, file))
        missing = df.isnull().sum()
        missing = missing[missing > 0]

        if not missing.empty:
            print(f"\n{file}")
            print(missing)
```

### **Duplicate Record Check**

Duplicate row counts across all datasets.

```python
for file in files:
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(data_path, file))
        duplicates = df.duplicated().sum()

        print(f"{file}: {duplicates} duplicate rows")
```

### **Geolocation Uniqueness Check**

Unique ZIP-code prefixes and total records in the geolocation dataset.

```python
geo_df = pd.read_csv(os.path.join(data_path, "olist_geolocation_dataset.csv"))

print("Total records:", len(geo_df))
print("Unique ZIP prefixes:", geo_df["geolocation_zip_code_prefix"].nunique())
print("Exact duplicate rows:", geo_df.duplicated().sum())
```

### **Order Status Distribution**

Count of orders by their recorded status.

```python
orders_df = pd.read_csv(os.path.join(data_path, "olist_orders_dataset.csv"))

print(orders_df["order_status"].value_counts())
```

### **Payment Method Distribution**

Count of transactions by payment type.

```python
payments_df = pd.read_csv(os.path.join(data_path, "olist_order_payments_dataset.csv"))

print(payments_df["payment_type"].value_counts())
```

### **Customer Review Distribution**

Count of customer reviews by rating score.

```python
reviews_df = pd.read_csv(os.path.join(data_path, "olist_order_reviews_dataset.csv"))

print(reviews_df["review_score"].value_counts().sort_index())
```

### **Product Category Distribution**

Count of products available in each category.

```python
products_df = pd.read_csv(os.path.join(data_path, "olist_products_dataset.csv"))

print(products_df["product_category_name"].value_counts().head(10))
```

### **Order Date Range**

Earliest and latest purchase dates recorded in the orders dataset.

```python
orders_df = pd.read_csv(os.path.join(data_path, "olist_orders_dataset.csv"))

orders_df["order_purchase_timestamp"] = pd.to_datetime(
    orders_df["order_purchase_timestamp"]
)

print("First order date:", orders_df["order_purchase_timestamp"].min())
print("Last order date:", orders_df["order_purchase_timestamp"].max())
```

# **Data Profiling**

### **Order ID Uniqueness Check**

Total order records compared with unique order IDs.

```python
orders_df = pd.read_csv(os.path.join(data_path, "olist_orders_dataset.csv"))

print("Total order records:", len(orders_df))
print("Unique order IDs:", orders_df["order_id"].nunique())
print("Duplicate order IDs:", orders_df["order_id"].duplicated().sum())
```

### **Customer ID Uniqueness Check**

Total customer records compared with unique customer IDs.

```python
customers_df = pd.read_csv(os.path.join(data_path, "olist_customers_dataset.csv"))

print("Total customer records:", len(customers_df))
print("Unique customer IDs:", customers_df["customer_id"].nunique())
print("Duplicate customer IDs:", customers_df["customer_id"].duplicated().sum())
```

### **Customer Identity Analysis**

Comparison of customer records and distinct customer identities.

```python
print("Unique customer IDs:", customers_df["customer_id"].nunique())
print("Unique customer identities:", customers_df["customer_unique_id"].nunique())
```

### **Repeat Customer Analysis**

Count of customers associated with multiple customer records.

```python
customer_orders = customers_df["customer_unique_id"].value_counts()

repeat_customers = (customer_orders > 1).sum()

print("Total unique customers:", customer_orders.nunique())
print("Repeat customers:", repeat_customers)
print("One-time customers:", (customer_orders == 1).sum())
```

### **Repeat Customer Analysis**

Identify one-time and repeat customers using unique customer identities.

```python
customer_orders = customers_df["customer_unique_id"].value_counts()

total_customers = len(customer_orders)
repeat_customers = (customer_orders > 1).sum()
one_time_customers = (customer_orders == 1).sum()

print("Total unique customers:", total_customers)
print("Repeat customers:", repeat_customers)
print("One-time customers:", one_time_customers)
```

### **Product ID Uniqueness Check**

Total product records compared with unique product IDs.

```python
products_df = pd.read_csv(os.path.join(data_path, "olist_products_dataset.csv"))

print("Total product records:", len(products_df))
print("Unique product IDs:", products_df["product_id"].nunique())
print("Duplicate product IDs:", products_df["product_id"].duplicated().sum())
```

### **Seller ID Uniqueness Check**

Total seller records compared with unique seller IDs.

```python
sellers_df = pd.read_csv(os.path.join(data_path, "olist_sellers_dataset.csv"))

print("Total seller records:", len(sellers_df))
print("Unique seller IDs:", sellers_df["seller_id"].nunique())
print("Duplicate seller IDs:", sellers_df["seller_id"].duplicated().sum())
```

### **Order Item Identifier Check**

Check whether each order and item combination is unique.

```python
items_df = pd.read_csv(os.path.join(data_path, "olist_order_items_dataset.csv"))

print("Total order items:", len(items_df))
print("Unique order-item combinations:",
      items_df.drop_duplicates(["order_id", "order_item_id"]).shape[0])
print("Duplicate combinations:",
      items_df.duplicated(["order_id", "order_item_id"]).sum())
```

### **Payment Record Analysis**

Compare payment records with distinct order IDs.

```python
payments_df = pd.read_csv(os.path.join(data_path, "olist_order_payments_dataset.csv"))

print("Total payment records:", len(payments_df))
print("Unique order IDs:", payments_df["order_id"].nunique())
print("Orders with multiple payment records:",
      (payments_df["order_id"].value_counts() > 1).sum())
```

### **Payment and Order Relationship Check**

Identify payment order IDs that do not exist in the orders dataset.

```python
order_ids = set(orders_df["order_id"])
payment_order_ids = set(payments_df["order_id"])

unmatched_payments = payment_order_ids - order_ids

print("Unmatched payment order IDs:", len(unmatched_payments))
```

### **Payment Identifier Validation**

Check whether order ID and payment sequence form a unique key.

```python
print("Unique payment combinations:",
      payments_df[["order_id", "payment_sequential"]].drop_duplicates().shape[0])

print("Total payment records:", len(payments_df))
```

### **Review Identifier Validation**

Check the uniqueness of review IDs.

```python
reviews_df = pd.read_csv(os.path.join(data_path, "olist_order_reviews_dataset.csv"))

print("Total review records:", len(reviews_df))
print("Unique review IDs:", reviews_df["review_id"].nunique())
print("Duplicate review IDs:", reviews_df["review_id"].duplicated().sum())
```

### **Duplicate Review Investigation**

Examine records sharing the same review ID.

```python
duplicate_reviews = reviews_df[
    reviews_df["review_id"].duplicated(keep=False)
]

print("Rows with duplicate review IDs:", len(duplicate_reviews))
print("Distinct duplicated review IDs:",
      duplicate_reviews["review_id"].nunique())

display(duplicate_reviews.head(5))
```

### **Repeated Review Comparison**

Compare records sharing the same review ID.

```python
review_id_counts = reviews_df["review_id"].value_counts()

most_repeated_id = review_id_counts.index[0]

display(
    reviews_df[reviews_df["review_id"] == most_repeated_id]
)
```

### **Review Composite Key Validation**

Check whether review ID and order ID uniquely identify each review record.

```python
print("Unique review-order combinations:",
      reviews_df[["review_id", "order_id"]].drop_duplicates().shape[0])

print("Total review records:", len(reviews_df))
```

### **Review and Order Relationship Check**

Identify review order IDs that do not exist in the orders dataset.

```python
order_ids = set(orders_df["order_id"])
review_order_ids = set(reviews_df["order_id"])

unmatched_reviews = review_order_ids - order_ids

print("Unmatched review order IDs:", len(unmatched_reviews))
```

### **Product and Order Item Relationship Check**

Identify order-item product IDs that do not exist in the products dataset.

```python
order_items_df = pd.read_csv(os.path.join(data_path, "olist_order_items_dataset.csv"))

product_ids = set(products_df["product_id"])
item_product_ids = set(order_items_df["product_id"])

unmatched_products = item_product_ids - product_ids

print("Unmatched product IDs:", len(unmatched_products))
```

### **Seller and Order Item Relationship Check**

Identify order-item seller IDs that do not exist in the sellers dataset.

```python
seller_ids = set(sellers_df["seller_id"])
item_seller_ids = set(order_items_df["seller_id"])

unmatched_sellers = item_seller_ids - seller_ids

print("Unmatched seller IDs:", len(unmatched_sellers))
```
